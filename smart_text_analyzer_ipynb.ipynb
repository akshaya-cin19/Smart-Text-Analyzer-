```python
# ============================================
# PROJECT 1: SMART TEXT ANALYZER
# Text and Speech Analysis
# ============================================

import re
from collections import Counter

print("=" * 55)
print("        📝 SMART TEXT ANALYZER")
print("=" * 55)

# Get input from user
text = input("\nEnter your paragraph:\n\n")

# Check empty input
if not text.strip():
    print("\n⚠️ Please enter some text!")
else:

    # -----------------------------
    # TEXT PROCESSING
    # -----------------------------

    # Extract words
    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    # Extract sentences
    sentences = re.split(r'[.!?]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    # Word frequency
    word_frequency = Counter(words)

    # Unique words
    unique_words = set(words)

    # Average word length
    if words:
        average_length = sum(len(word) for word in words) / len(words)
    else:
        average_length = 0

    # -----------------------------
    # DISPLAY RESULTS
    # -----------------------------

    print("\n" + "=" * 55)
    print("             📊 TEXT ANALYSIS")
    print("=" * 55)

    print(f"\n🔤 Characters       : {len(text)}")
    print(f"📝 Words            : {len(words)}")
    print(f"📌 Sentences        : {len(sentences)}")
    print(f"🔑 Unique Words     : {len(unique_words)}")
    print(f"📏 Average Word Len : {average_length:.2f}")

    # -----------------------------
    # MOST COMMON WORDS
    # -----------------------------

    print("\n" + "-" * 55)
    print("           🔥 TOP 10 COMMON WORDS")
    print("-" * 55)

    if word_frequency:
        for word, count in word_frequency.most_common(10):
            print(f"{word:<20} → {count} time(s)")
    else:
        print("No valid words found.")

    print("\n" + "=" * 55)
    print("        ✅ ANALYSIS COMPLETED")
    print("=" * 55)
```
